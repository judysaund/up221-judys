```python
# Rent burden variables based on your naming conventions
rent_burden_col = "Rent_Burden_30+"  # Modify if needed (e.g., "Rent_Burden_50+")
```

```python
# Merge 2007 and 2023 rent burden data using FIPS
merged_df = Y2007[['FIPS', rent_burden_col]].merge(
    Y2023[['FIPS', rent_burden_col]], on="FIPS", suffixes=('_2007', '_2023')
)
```

```python
# Compute the percentage change in rent burden
merged_df["Rent_Burden_Change"] = (
    (merged_df[f"{rent_burden_col}_2023"] - merged_df[f"{rent_burden_col}_2007"]) / 
    merged_df[f"{rent_burden_col}_2007"]
) * 100
```

```python
# Load Census Tract geometries (Assumes you have 'census_tracts_LA.geojson')
tracts = gpd.read_file('2023 Census Tracts.zip')
```

```python
#subset caTracts where CountyFP equals '037'
la_tracts = tracts[tracts['COUNTYFP']=='037']
la_tracts.head()
```

```python
# Merge with existing LA census tracts (la_tracts)
la_tracts = la_tracts.merge(merged_df, left_on="GEOID", right_on="FIPS", how="left")

# DEBUG: Ensure Rent_Burden_Change exists after merging
print("Columns after merging with la_tracts:", la_tracts.columns)
```

```python
# we only really need FIPS and geometry, so let's subset the data
la_tracts = tracts[['NAMELSAD', 'TRACTCE', 'GEOID', 'geometry']]
la_tracts.head()
```

```python
# Merge rent burden change data with LA census tracts
la_tracts = la_tracts.merge(merged_df, left_on="GEOID", right_on="FIPS", how="left")

# DEBUG: Ensure Rent_Burden_Change exists after merging
print("Columns after merging with la_tracts:", la_tracts.columns)
```

```python
# Load LA Neighborhood Boundaries from ArcGIS API
arcgis_url = "https://services5.arcgis.com/7nsPwEMP38bSkCjy/arcgis/rest/services/LA_Times_Neighborhoods/FeatureServer/0/query?outFields=*&where=1%3D1&f=geojson"
neighborhoods = gpd.read_file(arcgis_url)
```

```python
# Ensure CRS matches for spatial join
la_tracts = la_tracts.to_crs(neighborhoods.crs)
```

```python
# Spatial Join: Assign each census tract to a neighborhood
tracts_with_neighborhoods = gpd.sjoin(la_tracts, neighborhoods, how="left", predicate="intersects")

# DEBUG: Ensure Rent_Burden_Change exists after spatial join
print("Columns after spatial join:", tracts_with_neighborhoods.columns)# Reassign Rent_Burden_Change in case it was dropped during spatial join
if "Rent_Burden_Change" not in tracts_with_neighborhoods.columns:
    tracts_with_neighborhoods["Rent_Burden_Change"] = la_tracts["Rent_Burden_Change"]
```

```python
# Reassign Rent_Burden_Change in case it was dropped during spatial join
if "Rent_Burden_Change" not in tracts_with_neighborhoods.columns:
    tracts_with_neighborhoods["Rent_Burden_Change"] = la_tracts["Rent_Burden_Change"]
```

```python
# Aggregate rent burden change at the neighborhood level using dissolve()
neighborhood_rent_burden = tracts_with_neighborhoods.dissolve(
    by="name", aggfunc={"Rent_Burden_Change": "mean"}
)
```

```python
# Ensure all geometries are valid before saving
neighborhood_rent_burden = neighborhood_rent_burden[neighborhood_rent_burden.is_valid]
```

```python
# Save the results as a GeoJSON file
neighborhood_rent_burden.to_file("rent_burden_change_neighborhoods.geojson", driver="GeoJSON")
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

# Load the final dataset
neighborhood_rent_burden = gpd.read_file("rent_burden_change_neighborhoods.geojson")

# Convert CRS to match LA's standard projection (EPSG:2229 or 4326)
neighborhood_rent_burden = neighborhood_rent_burden.to_crs(epsg=2229)  # EPSG:2229 = California State Plane (LA)

# Set up the figure and axis
fig, ax = plt.subplots(figsize=(12, 10))

# Ensure correct aspect ratio to prevent stretching
ax.set_aspect(1)

# Create a colormap
cmap = plt.cm.RdBu_r  # Red for increase, Blue for decrease
norm = mcolors.Normalize(vmin=neighborhood_rent_burden["Rent_Burden_Change"].min(),
                         vmax=neighborhood_rent_burden["Rent_Burden_Change"].max())

# Plot the neighborhoods with rent burden change as a heatmap
neighborhood_rent_burden.plot(column="Rent_Burden_Change",
                              cmap=cmap,
                              linewidth=0.6,
                              edgecolor="black",
                              norm=norm,
                              legend=True,
                              legend_kwds={"label": "Change in Rent Burden (%)",
                                           "orientation": "vertical"},
                              ax=ax)

# Add a title
ax.set_title("Heatmap of Rent Burden Change by Neighborhood (2007-2023)", fontsize=14)

# Remove axes for a cleaner map
ax.set_xticks([])
ax.set_yticks([])
ax.set_frame_on(False)

# Show the plot
plt.show()
```
